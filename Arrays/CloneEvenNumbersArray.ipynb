```
Given an array of numbers, replace each even number with two
of the same number. e.g, [1,2,5,6,8, , , ,] -> [1,2,2,5,6,6,8,8].
Assume that the array has the exact amount of space to accommodate the result.
```

```
public static int[] CloneEvenNumbers(int[] a) {
    if (a == null || a.Length == 0)
    {
        return a;
    }
    int end = a.Length, i = getLastNumberIndex(a);
    while (i >=0)
    {
        if(a[i]%2 == 0)
        {
            a[--end] = a[i];
        }
        a[--end] = a[i];
        i--;
    }
    return a;
}

public static int getLastNumberIndex(int[] a)
{
    int i = a.Length-1;
    while (i>=0 && a[i] ==-1){
        i--;
    }
    return i;
}

//TEST
int[] a = [2,5,6,8,-1,-1,-1];
CloneEvenNumbers(a)
```

```
Given an array of numbers, replace each even number with two of the same number. 
e.g, [1,2,5,6,8] -> [1,2,2,5,6,6,8,8]. Assume that the array has enough space to accommodate the result.
```

```
public static int[] CloneEvenNumbers(int[] a)
{
    if (a == null || a.Length == 0)
    {
        return a;
    }
    int end = a.Length;
    int i = getLastNumIndex(a);

    while(i >= 0)
    {
        if(a[i]%2 == 0)
        {
            end = end - 1;
            end = a[i];
        }

        end = a[i];
        i--;
    }
    return a;
}

public static int getLastNumIndex(int[] a)
{
    int i = a.Length-1;
    while(i >= 0 && a[i] == -1)
    {
        i--;
    }
    return i;
}

//TEST
int[] a = [2,5,6,8,-1,-1,-1];
CloneEvenNumbers(a)
```
